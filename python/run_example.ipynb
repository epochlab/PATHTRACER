```python
from dataclasses import replace
import matplotlib.pyplot as plt

from pathtracer import Renderer, display_encode
```

```python
renderer = Renderer("scenes/cornell.json")
```

```python
cam = replace(renderer.default_camera,
              lens="fisheye_polynomial",
              position=(0.0, 0.0, 0.5),
              yaw_degrees=0.0,
              pitch_degrees=0.0,
              film_back_mm=(36.0, 24.0),
              focal_length_mm=8.0,
              near_clip=0.01,
              far_clip=100.0,
              aperture=2.8,
              shutter_seconds=0.008,
              iso=400.0)

frame = renderer.render(camera=cam, 
                        aovs=("beauty",), 
                        width=256, height=128, 
                        samples=16, 
                        seed=1, 
                        show_sky=False, 
                        env_light_enabled=True)
```

```python
x = frame["beauty"]
print(f"Min: {x.min()}, Max: {x.max()}")
```

```python
plt.imshow(display_encode(x))
plt.xticks([]), plt.yticks([])
plt.tight_layout()
```
