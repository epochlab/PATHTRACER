```python
from dataclasses import replace
import matplotlib.pyplot as plt

from pathtracer import Renderer, display_encode, lattice_preview
```

```python
scene = Renderer("scenes/cornell.json")

camera = replace(scene.default_camera)

frame = scene.render(camera=camera, aovs=("beauty",))
```

```python
x = frame["beauty"]
print(f"Min: {x.min()}, Max: {x.max()}")
```

```python
# plt.imshow(display_encode(x, camera=cam))
plt.imshow(display_encode(lattice_preview(x, camera, scale=8), camera=camera))
plt.xticks([]), plt.yticks([])
plt.tight_layout()
```
